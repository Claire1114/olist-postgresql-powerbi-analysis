```python
import sqlite3
import pandas as pd
from sqlalchemy import create_engine

# 1. 連到 SQLite 檔案
sqlite_conn = sqlite3.connect('/Users/claireweng/Documents/postgresSQL/olist.sqlite')

# 2. 設定 PostgreSQL 資訊 
pg_engine = create_engine('postgresql://localhost/postgres') 

# 3. 獲取 SQLite 裡面所有的資料表名稱
cursor = sqlite_conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = [row[0] for row in cursor.fetchall()]

# 4. 把每一張表用 Pandas 讀出來，再寫進 PostgreSQL
for table_name in tables:
    print(f"正在搬運資料表: {table_name}...")
    df = pd.read_sql_query(f"SELECT * FROM {table_name}", sqlite_conn)
    
    # 寫入 PostgreSQL，如果表存在就替換掉
    df.to_sql(table_name, pg_engine, if_exists='replace', index=False)

print("🎉 所有資料表已成功匯入 PostgreSQL！")

# 關閉連線
sqlite_conn.close()
```
